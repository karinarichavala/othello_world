# Visualización del Primer Tablero Reconstruido

Este notebook muestra el primer tablero reconstruido a partir de las partidas sintéticas guardadas.

```python
import numpy as np
```

```python
# Cargar tableros reconstruidos
boards = np.load('../metrics/data/board_states_200games.npy')
print(f"Shape de los tableros: {boards.shape}")
print(f"Tipo de datos: {boards.dtype}")
print(f"Valores únicos: {np.unique(boards)}")
```

```python
# Leer los movimientos de la primera partida
with open('../activations/data/games_200.txt', 'r') as f:
    first_game_moves = f.readline().strip().split()

print(f"Movimientos de la primera partida:")
print(f"Total de movimientos: {len(first_game_moves)}")
print(f"Primeros 10 movimientos: {first_game_moves[:10]}")
```

```python
def visualize_board(board, move_num=None, move_pos=None):
    """Visualiza un tablero de Othello de forma legible"""
    if move_num is not None:
        print(f"\n{'='*50}")
        print(f"DESPUÉS DEL MOVIMIENTO {move_num}")
        if move_pos is not None:
            row, col = int(move_pos) // 8, int(move_pos) % 8
            print(f"Movimiento jugado en: casilla {move_pos} → posición ({row}, {col}) = {chr(65+col)}{row+1}")
        print(f"{'='*50}\n")
    
    print("  A B C D E F G H")
    for i, row in enumerate(board):
        symbols = []
        for val in row:
            if val == 1:
                symbols.append('●')  # Negro
            elif val == -1:
                symbols.append('○')  # Blanco
            else:
                symbols.append('·')  # Vacío
        print(f"{i+1} {' '.join(symbols)}")
    
    # Contar piezas
    black_count = np.sum(board == 1)
    white_count = np.sum(board == -1)
    empty_count = np.sum(board == 0)
    
    print(f"\nNegras (●): {black_count}")
    print(f"Blancas (○): {white_count}")
    print(f"Vacías (·): {empty_count}")
```

## Primer Tablero de la Primera Partida

```python
# Mostrar el primer tablero (después del primer movimiento)
first_board = boards[0, 0]
first_move = first_game_moves[0]

visualize_board(first_board, move_num=1, move_pos=first_move)
```

## Primeros 5 Movimientos de la Partida

```python
# Mostrar el tablero final después de todos los movimientos
final_board = boards[0, -1]  # Último movimiento (movimiento 59)
final_move = first_game_moves[-1]

print(f"Total de movimientos en la partida: {len(first_game_moves)}")
visualize_board(final_board, move_num=len(first_game_moves), move_pos=final_move)
```

## Verificar el Movimiento 60

```python
# Verificar si la primera partida tiene 60 movimientos en el archivo original
print(f"Movimientos en el archivo original: {len(first_game_moves)}")

# Leer todas las partidas para ver la distribución
with open('../activations/data/games_200.txt', 'r') as f:
    all_lines = f.readlines()

lengths = [len(line.strip().split()) for line in all_lines]
print(f"\nDistribución de longitudes:")
print(f"  Partidas con 60 movimientos: {lengths.count(60)}")
print(f"  Partidas con 59 movimientos: {lengths.count(59)}")

if len(first_game_moves) == 60:
    print(f"\n✓ La primera partida SÍ tiene movimiento 60: {first_game_moves[59]}")
    print(f"  Pero se truncó a 59 porque el modelo usa block_size=59")
```
