# Visualización de Ground Truth BSPs

Este notebook muestra el contenido de los archivos de ground truth:
- `bsp_ground_truth_200games.npy`: Matriz de datos (11800, 198)
- `bsp_ground_truth_200games.names.npy`: Nombres de las columnas (198,)

```python
import numpy as np
from pathlib import Path
import pandas as pd

# Proyecto root
project_root = Path('../..').resolve()
print(f"Proyecto: {project_root}")
```

## 1. Cargar Archivos

```python
# Cargar ground truth (datos)
bsp_gt_path = project_root / "sae" / "metrics" / "data" / "bsp_ground_truth_200games.npy"
bsp_ground_truth = np.load(bsp_gt_path)

print(" bsp_ground_truth_200games.npy")
print("=" * 60)
print(f"Shape: {bsp_ground_truth.shape}")
print(f"Dtype: {bsp_ground_truth.dtype}")
print(f"Valores únicos: {np.unique(bsp_ground_truth)}")
print(f"Memoria: {bsp_ground_truth.nbytes / (1024**2):.2f} MB")
print(f"\nInterpretación:")
print(f"  - 11,800 posiciones (200 partidas × 59 movimientos)")
print(f"  - 198 BSPs por posición")
print(f"  - Valores: True (1) o False (0)")
```

```python
# Cargar nombres (metadatos)
bsp_names_path = project_root / "sae" / "metrics" / "data" / "bsp_ground_truth_200games.names.npy"
bsp_names = np.load(bsp_names_path, allow_pickle=True)

print(" bsp_ground_truth_200games.names.npy")
print("=" * 60)
print(f"Shape: {bsp_names.shape}")
print(f"Dtype: {bsp_names.dtype}")
print(f"Tipo de elementos: {type(bsp_names[0])}")
print(f"\nInterpretación:")
print(f"  - 198 nombres (uno por cada columna del archivo anterior)")
print(f"  - Formato string: 'BSPA10', 'BSPA11', etc.")
```

## 2. Ver Primeras y Últimas BSPs

```python
print(" Primeras 20 BSPs:")
print("=" * 60)
for i in range(20):
    print(f"Columna {i:3d}: {bsp_names[i]:15s}")
```

```python
print(" Últimas 10 BSPs (especiales):")
print("=" * 60)
for i in range(len(bsp_names) - 10, len(bsp_names)):
    print(f"Columna {i:3d}: {bsp_names[i]:15s}")
```

### 3. Ejemplo: Ver Una Posición Completa

```python
# Analizar la posición 0 (primer movimiento de la primera partida)
posicion_idx = 0
posicion_data = bsp_ground_truth[posicion_idx]

print(f" Posición {posicion_idx} (Partida 0, Movimiento 0):")
print("=" * 60)
print(f"Total de BSPs activas: {np.sum(posicion_data)}")
print(f"Total de BSPs inactivas: {len(posicion_data) - np.sum(posicion_data)}")
print()

# Mostrar BSPs activas por categoría
print(" BSPs activas (True):")
print()

vacias_activas = []
mias_activas = []
oponente_activas = []
especiales_activas = []

for i, (name, valor) in enumerate(zip(bsp_names, posicion_data)):
    if valor:
        if len(name) == 6:
            if name.endswith('0'):
                vacias_activas.append(name)
            elif name.endswith('1'):
                mias_activas.append(name)
            elif name.endswith('2'):
                oponente_activas.append(name)
        elif name.startswith('BSP_'):
            especiales_activas.append(name)

print(f"Casillas vacías ({len(vacias_activas)}):")
if vacias_activas:
    casillas_vacias = [name[3:5] for name in vacias_activas]
    print(f"  {', '.join(casillas_vacias[:20])}")
    if len(casillas_vacias) > 20:
        print(f"  ... y {len(casillas_vacias) - 20} más")

print(f"\nMis piezas ({len(mias_activas)}):")
if mias_activas:
    casillas_mias = [name[3:5] for name in mias_activas]
    print(f"  {', '.join(casillas_mias)}")

print(f"\nPiezas oponente ({len(oponente_activas)}):")
if oponente_activas:
    casillas_oponente = [name[3:5] for name in oponente_activas]
    print(f"  {', '.join(casillas_oponente)}")

print(f"\nEstados especiales ({len(especiales_activas)}):")
for name in especiales_activas:
    print(f"  ✓ {name}")
```

### 4. Crear DataFrame para Visualización

```python
# Crear DataFrame de las primeras 10 posiciones
n_show = 10
df = pd.DataFrame(
    bsp_ground_truth[:n_show],
    columns=bsp_names,
    index=[f"Pos_{i}" for i in range(n_show)]
)

print(f" DataFrame de primeras {n_show} posiciones:")
print("=" * 60)
print(df.iloc[:, :10])  # Mostrar solo primeras 10 columnas
print(f"\n... (198 columnas total, mostrando solo las primeras 10)")
```
