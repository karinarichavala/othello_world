```python
import numpy as np
import matplotlib.pyplot as plt

data = np.load('reconstruction_results.npz', allow_pickle=True)
print('Keys:', list(data.keys()))
```

```python
reconstruction_score = data['reconstruction_score'].item()
best_threshold       = data['best_threshold'].item()
f1_per_threshold     = data['f1_per_threshold']
thresholds           = data['thresholds']
phase1_time          = data['phase1_time_seconds'].item()
phase2_time          = data['phase2_time_seconds'].item()
total_time           = data['total_time_seconds'].item()
bsp_names            = data['bsp_names']

print(f'Reconstruction score : {reconstruction_score:.4f}')
print(f'Best threshold       : {best_threshold:.4f}')
print(f'Phase 1 time         : {phase1_time:.2f} s')
print(f'Phase 2 time         : {phase2_time:.2f} s')
print(f'Total time           : {total_time:.2f} s')
print(f'BSP names ({len(bsp_names)})       : {bsp_names[:10]} ...')
```

```python
plt.figure(figsize=(7, 4))
plt.plot(thresholds, f1_per_threshold, marker='o')
plt.axvline(best_threshold, color='red', linestyle='--', label=f'Best threshold = {best_threshold:.3f}')
plt.xlabel('Threshold')
plt.ylabel('F1 score')
plt.title('F1 per threshold')
plt.legend()
plt.tight_layout()
plt.show()
```
